## 1 — HALLAZGOS DEL ANÁLISIS

A lo largo del proyecto obtuvimos los siguientes hallazgos con evidencia:

**Sobre los planes de suscripción:**
El plan Básico concentra la mayor parte de los usuarios de la plataforma. El plan Premium tiene la menor adopción pero sus usuarios muestran el mayor tiempo de visualización mensual, tanto en media como en mediana. Esto sugiere que el nivel de consumo está asociado al tipo de plan contratado.

**Sobre las edades:**
La distribución de edades es aproximadamente uniforme entre los usuarios, sin concentración en un grupo etario particular. Esto indica que la plataforma tiene alcance transversal a distintas generaciones.

**Sobre los géneros y países:**
La preferencia de géneros es relativamente homogénea entre países, con pequeñas variaciones culturales. No existe un género dominante que se imponga claramente sobre los demás en ningún país analizado.

**Sobre la relación edad y consumo:**
No existe correlación lineal fuerte entre la edad del usuario y el tiempo mensual de visualización. Sin embargo, al incorporar el plan de suscripción como tercera variable, se observa que los usuarios Premium tienden a concentrarse en niveles de consumo más altos independientemente de la edad.

**Sobre el PCA:**
Las tres variables numéricas analizadas (edad, tiempo de visualización y tickets de soporte) muestran contribuciones diferenciadas en las componentes principales. La primera componente captura la mayor proporción de varianza del dataset.

## 2 — LIMITACIONES

- El dataset contiene solo 3 variables numéricas continuas, lo que acota el alcance del PCA y del análisis multivariado.
- Las variables categóricas (plan, país, género) no fueron incorporadas al PCA por requerir codificación adicional, lo que limita la riqueza del análisis de reducción de dimensionalidad.
- Los valores faltantes y los errores de carga presentes en el dataset original pueden haber introducido sesgos en las distribuciones, aun después de la limpieza realizada.
- El alcance de las conclusiones se encuentra condicionado por la información disponible y por las decisiones documentadas durante el proceso.

## 3 — MEJORAS FUTURAS

- Incorporar la codificación de variables categóricas (plan, país, género) para enriquecer el análisis de PCA y obtener componentes más interpretables.
- Aplicar técnicas de clustering para identificar grupos de usuarios con comportamientos similares de consumo.
- Incorporar variables adicionales como frecuencia de login o historial de cambios de plan para ampliar el análisis exploratorio.
- Explorar modelos predictivos para estimar el plan de suscripción o el tiempo de visualización a partir de las variables disponibles.
