```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_json('../data/raw/streaming_users_dirty.csv')
df.head()
```

## ESTRUCTURA GENERAL DEL DATASET

Verificamos cuántas filas y columnas tiene el dataset y los nombres de cada columna.

```python
print(f"Filas: {df.shape[0]}")
print(f"Columnas: {df.shape[1]}")
print(f"\nNombre de las columnas: {list(df.columns)}")
```

```python
print("Tipos de datos:")
print(df.dtypes)
```

Observamos que:
- `user_id` → int64 (número entero)
- `age` → int64 (número entero)
- `subscription_plan` → object (texto)
- `monthly_watch_time_mins` → float64 (decimal)
- `country` → object (texto)
- `favorite_genre` → object (texto)
- `last_login_date` → object (texto) — debería ser de tipo fecha
- `customer_support_tickets` → int64 (número entero)

```python
print("Valores faltantes por columna:")
print(df.isnull().sum())
print(f"\nTotal de valores faltantes: {df.isnull().sum().sum()}")
```

Encontramos 753 valores faltantes en total distribuidos así:
- `monthly_watch_time_mins` → 193 nulos
- `favorite_genre` → 240 nulos
- `last_login_date` → 320 nulos

```python
duplicados = df.duplicated().sum()
print(f"Registros duplicados: {duplicados}")
```

Hay 126 registros duplicados. Eso significa que 126 filas aparecen más de una vez con exactamente los mismos datos.

```python
print("Valores únicos en subscription_plan:")
print(df['subscription_plan'].unique())

print("\nValores únicos en country:")
print(df['country'].unique())

print("\nValores únicos en favorite_genre:")
print(df['favorite_genre'].unique())
```

Observamos que:
- `subscription_plan` tiene 15 variantes para solo 3 planes reales: Básico, Estándar y Premium escritos con mayúsculas, minúsculas, acentos, abreviaciones y errores tipográficos.
- `country` tiene países escritos con nombres completos, abreviados, en minúsculas y con espacios extra.
- `favorite_genre` tiene géneros en español e inglés mezclados, con mayúsculas, minúsculas y errores tipográficos.

```python
df[['age', 'monthly_watch_time_mins', 'customer_support_tickets']].describe()
```

Observamos valores imposibles en las tres columnas numéricas:
- `age` → mínimo -5 y máximo 150, ambos imposibles para una edad real.
- `monthly_watch_time_mins` → mínimo -120 (tiempo negativo) y máximo 99999 (más de 69 días seguidos, claramente un error).
- `customer_support_tickets` → mínimo -1 (tickets negativos) y máximo 150 (valor extremadamente alto).

## OBSERVACIONES FINALES

A partir de la inspección identificamos los siguientes problemas en el dataset:

- **Dimensiones:** 8.160 filas y 8 columnas.
- **Valores faltantes:** 753 en total. Columnas afectadas: `monthly_watch_time_mins` (193), `favorite_genre` (240) y `last_login_date` (320).

- **Tipo de dato incorrecto:** `last_login_date` figura como texto cuando debería ser fecha.
